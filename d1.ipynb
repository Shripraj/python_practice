```python
def remove_duplicates(lst):
    list1=list()
    lst=lst
    for i in lst:
        if i not in list1:
            list1.append(i)
        else:
            i+=1
    return list1
lst=[1, 2, 3, 4, 5, 1, 2]
remove_duplicates(lst)
```
