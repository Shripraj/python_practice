```python
n=int(input("Enter a digit:"))
i=1
count=0
while i<=n:
    count+=1
    i+=1
print(count)
```

```python
n=int(input("Enter the number:"))
reversed=0
while n>0:
    b=n%10
    reversed=(reversed*10)+b
    n=n//10
print(reversed)
```

```python
n=input("Enter the number:")
right=0
left=len(n)-1
is_palindrome=True
while right<left:
    if n[right] != n[left]:
        is_palindrome=False
        break
    right=right+1
    left=left-1
if is_palindrome:
    print("is a palindrome")
else:
    print("is not a palindrome")
```
