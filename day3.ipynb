```python
# 2
def nsum(n):
    tsum=0
    for i in range(1,n+1):
        tsum +=i
    return tsum 
n=int(input("Enter a number:"))
nsum(n)
```

```python
# 3
def tabels(num):
    for i in range(1,11):
        print(f"{num}x{i}={num*i}")
num=int(input("Enter the number:"))
tabels(num)
```

```python
# 4
def odd_or_even(nums):
    n=nums
    print(f"List is {n}")
    even=0
    odd=0
    for i in n:
        if i%2==0:
            even+=1
            i=i+1
        else:
            odd+=1
            i=i+1
    print(f"even={even}")
    print(f"odd={odd}")
nums=list(map(int,input("Enter the numbers").split()))
odd_or_even(nums)
```

```python
# 6
def palindrome(n):
    num =n
    result=0
    while num>0:
        id=num%10
        result=(result*10)+id
        num=num//10
    return n == result
palindrome(121)
```

```python
#7
def fact(n):
    r=1
    while n>0:
        r=r*n
        n=n-1
    return r
fact(6)
```

```python
prime_list=[]
n=int(input("Enter a start number:"))
m=int(input("Enter a end number:"))
for i in range(n,m):
    if i < 0:
        continue
    elif i==1 or i==2:
        prime_list.append(i)
    elif i%2!=0 and i%3!=0:
        prime_list.append(i)
print(prime_list)
    
```
