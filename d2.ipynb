```python
# 1. Positive, Negative, or Zero
num=int(input("Enter a number:"))
if num==0:
    print("the number is zero")
elif num>0:
    print("the number is positive")
else :
    print("the number is negative")
```

```python
# 4. Leap Year Checker
year=int(input("Enter the year:"))
if year %400==0:
    print(f"{year} is a leap year") 
else:
    print(f"{year} is not leap year")
```

```python
# 5. Student Grade Calculator
try:
    name=input("Enter the name of the student:")
    marks=int(input("Enter the marks of student:"))
    if marks < 0 or marks > 100:
        print("Error: Marks should be between 0 and 100.")
    elif marks >=90 and marks<=100:
        print(f"Excelent {name} your marks is {marks} and your grade is A+ !kuudoss! to you")
    elif marks >=80 and marks<=89:
        print(f"VeryGood {name} your marks is {marks} and your grade is A !kuudoss! to you")
    elif marks >=70 and marks<=79:
        print(f"Good {name} your marks is {marks} and your grade is B !kuudoss! to you")
    elif marks >=60 and marks<=69:
        print(f"Excelent {name} your marks is {marks} and your grade is C !kuudoss! to you")
    elif marks >=35 and marks<=59:
        print(f"Good {name} your marks is {marks} and your grade is D !kuudoss! to you")
    elif marks <35:
        print(f"{name}! you are failed \n need to work more marks are not one to decide your future its only you !")
except ValueError as e:
    print(e)
finally:
    print("Thank you")
```

```python
#6. BMI Category 
weight=float(input("Enter you weight in kgs:"))
height=float(input("Enter your height in meters:"))
bmi = weight / (height ** 2)
if bmi<18.5:
    print("under weight")
elif 24.9>=bmi>18.5:
    print("you are perfect")
elif 29.9>=bmi>25.0:
    print("over weight")
else:
    print("obase")

```
